```python
f1=open("myfile.txt","w")
L=["This is delhi\n","This is paris\n","This is london\n"]
f1.write("Hello\n")
f1.writelines(L)
f1.close()
f1=open("myfile.txt","r+")
print('output of read():')
print(f1.read())
print()
f1.seek(0)
print('output of readline():')
print(f1.readline())
print()
f1.seek(0)
print('output of read(9):')
print(f1.read(9))
print()
```
